<div style="background-color: #f6a800; color: #ffffff; padding: 10px;">
    <h2> Part 1 &mdash; caveman</h2>
</div>

**caveman** is a Claude Code skill that *compresses Claude's own output tokens*. Tagline:

> *"Why use many token when few token do trick."*

Claimed reduction: **65&ndash;75%** of output tokens while preserving technical accuracy. Repository: <https://github.com/JuliusBrussee/caveman> &middot; License: MIT.

<div style="background-color: #dd6108; color: #ffffff; padding: 10px;">
<h3>1. The problem it solves</h3>
</div>

LLM responses are verbose by default. Articles, hedging, pleasantries, and transitional phrases all cost tokens and reading time. Most of it is cognitive *texture* &mdash; nice for user-facing copy, wasteful in a pair-programming session where you're reading the transcript at speed.

Two concrete pains caveman addresses:

- **Cost.** Output tokens are billed. 70% fewer output tokens = 70% cheaper responses on long sessions.
- **Readability-at-speed.** When you're iterating at one-turn-per-minute, dense text is faster to parse.

<div style="background-color: #dd6108; color: #ffffff; padding: 10px;">
<h3>2. What it does</h3>
</div>

caveman activates a compressed output style. Three intensities:

| Mode | Style | Example response (same task) |
|------|-------|----------------------------|
| **Lite** | Professional, no fluff. Grammar intact. | *"The object reference changes every render. Inline-object props break memoisation. Wrap in `useMemo`."* |
| **Full** | Default. Fragments acceptable. | *"New object ref each render. Inline object prop = new ref = re-render. Wrap in useMemo."* (19 tokens vs. ~69 normal) |
| **Ultra** | Telegraphic. Maximum compression. | *"ref churn. inline object = new ref. useMemo fix."* |

Everything technical stays in (variable names, code, line numbers). Everything ceremonial gets cut.

<h3>3. Install</h3>
</div>

```bash
claude plugin marketplace add JuliusBrussee/caveman
claude plugin install caveman@caveman
```

Verify:

```bash
claude plugin list   # caveman@caveman should appear, Status: enabled
```

Cross-agent install (if you also use Cursor / Windsurf / Cline / Copilot / Codex / Gemini):

```bash
npx skills add JuliusBrussee/caveman -a cursor
# or:  -a windsurf, -a cline, -a copilot, -a codex, -a gemini
```

Standalone (no plugin host, bare hooks):

```bash
bash <(curl -s https://raw.githubusercontent.com/JuliusBrussee/caveman/main/hooks/install.sh)
```

<div style="background-color: #dd6108; color: #ffffff; padding: 10px;">
<h3>4. Activate</h3>
</div>

Two ways:

```text
/caveman
```

or just say:

```text
talk like caveman
```

Switch intensity mid-session:

```text
/caveman lite    # professional-terse
/caveman full    # default caveman
/caveman ultra   # telegraphic
/caveman off     # back to normal
```

The plugin adds a statusline badge so you can see at a glance whether caveman is active and which mode you're in.

<div style="background-color: #dd6108; color: #ffffff; padding: 10px;">
<h3>5. When to use</h3>
</div>

**Good fits:**

- Rapid iteration where you read the transcript at speed.
- Pair-programming on known territory &mdash; you only need the *delta* from the previous turn.
- Long debug sessions whose output cost is adding up.
- Teaching yourself by example: the compressed form often reads like a cheat-sheet.

**Poor fits:**

- **User-facing copy.** Marketing, release notes, error messages for end users &mdash; keep caveman off.
- **PR descriptions, commit messages, documentation.** Reviewers want complete sentences.
- **Onboarding a new engineer.** They need the ceremony; caveman omits it.
- **Any step where you'll *paste* Claude's output elsewhere.** Re-enable normal voice before asking for copy that will be consumed by humans who are not you.

> **Pattern:** turn caveman on at the start of a long debug session. Before the step that writes the PR description, turn it off.
