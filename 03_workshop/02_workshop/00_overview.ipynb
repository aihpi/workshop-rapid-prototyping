    <h2> Overview</h2>
</div>

Welcome to **Rapid Prototyping via Vibe Coding**.

"Vibe coding" is the practice of *describing what you want* and letting an agentic coding environment figure out *how to build it*.
You explore, plan, verify &mdash; the coding agent reads files, runs commands, writes tests, and shows you diffs.
The skill you develop in this workshop is **driving the agent well**: keeping it focused, giving it verification hooks, and stopping it before it goes off the rails.

Part 2 focuses on **Claude Code** as the foundational vibe-coding tool, and four complementary add-ons that make it faster and cheaper to work with.

<h3>3. Prerequisites</h3>
</div>

Before running any of the notebooks you need:

- A **terminal** on macOS or Linux (Windows works via WSL2).
- **Node.js &ge; 18** (`node --version`). Claude Code and context-mode both require it. Node 22+ recommended on Linux.
- **Python 3.11+** with either `uv` or `pip`. The docling-toolkit uses `uv`; GEPA installs via `pip`.
- **Git**, and a **project directory under version control** so you can see what Claude changes.
- An **Anthropic account** &mdash; you'll log in the first time you run `claude`.
- *(optional but recommended)* The **`gh` CLI** authenticated against GitHub. Claude Code works well with `gh` for PRs, issues, and gists.

You do **not** need an OpenAI key for the core Claude Code workflow. GEPA's example in `04_*` does talk to an LLM &mdash; you can point it at Anthropic, OpenAI, or any LiteLLM-compatible provider.

<h3>5. Environment check (optional)</h3>
</div>

The cell below prints the versions of the tools you'll need. Running it is optional &mdash; if anything is missing, the notebook it gates will tell you what to install.

```python
import shutil
import subprocess

def probe(cmd: list[str], label: str | None = None) -> None:
    label = label or cmd[0]
    if shutil.which(cmd[0]) is None:
        print(f"{label:<18} NOT FOUND")
        return
    try:
        out = subprocess.check_output(cmd, text=True, stderr=subprocess.STDOUT).strip().splitlines()[0]
    except subprocess.CalledProcessError as exc:
        out = f"error: {exc}"
    print(f"{label:<18} {out}")

probe(["node", "--version"])
probe(["npm", "--version"])
probe(["python3", "--version"])
probe(["git", "--version"])
probe(["gh", "--version"], "gh")
probe(["claude", "--version"])
```
